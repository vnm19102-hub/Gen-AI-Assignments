```python
vector_A = [2, 4, 1]
vector_B = [1, 3, 5]

print("Vector A:", vector_A)
print("Vector B:", vector_B)
def dot_product(A, B):
    total = 0
    for a, b in zip(A, B):
        total += a * b
    return total

dp = dot_product(vector_A, vector_B)
print("Dot Product (A . B):", dp)


import math

def magnitude(A):
    total = 0
    for a in A:
        total += a ** 2
    return math.sqrt(total)

mag_A = magnitude(vector_A)
mag_B = magnitude(vector_B)

print("Magnitude of A:", round(mag_A, 4))
print("Magnitude of B:", round(mag_B, 4))

def cosine_similarity_manual(A, B):
    dp = dot_product(A, B)
    mag_A = magnitude(A)
    mag_B = magnitude(B)
    similarity = dp / (mag_A * mag_B)
    return similarity

result_manual = cosine_similarity_manual(vector_A, vector_B)
print("Cosine Similarity (Manual Calculation):", round(result_manual, 4))


```

verifing using numpy


```python
import numpy as np

A = np.array(vector_A)
B = np.array(vector_B)

dot = np.dot(A, B)
norm_A = np.linalg.norm(A)
norm_B = np.linalg.norm(B)

result_numpy = dot / (norm_A * norm_B)
print("Cosine Similarity (NumPy):", round(result_numpy, 4))
```

# Write cosine_similarity_manual using list comprehensions instead of loops.

```python
import math

def cosine_similarity_manual(A, B):
    # Dot product using list comprehension
    dot_product = sum([a * b for a, b in zip(A, B)])

    # Magnitudes using list comprehension
    magnitude_A = math.sqrt(sum([a ** 2 for a in A]))
    magnitude_B = math.sqrt(sum([b ** 2 for b in B]))

    # Cosine similarity
    return dot_product / (magnitude_A * magnitude_B)


# Example
A = [1, 2, 3]
B = [4, 5, 6]

result = cosine_similarity_manual(A, B)

print("Cosine Similarity:", result)
```
